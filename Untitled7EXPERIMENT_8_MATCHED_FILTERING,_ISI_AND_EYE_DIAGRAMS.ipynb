```python
# ============================================================
# EXPERIMENT 8
# MATCHED FILTERING, ISI AND EYE DIAGRAMS
# BPSK + RRC + AWGN + MULTIPATH ISI
# Google Colab / Python
# No Communication Toolbox required
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

# ============================================================
# 1. PARAMETERS
# ============================================================

N = 10000
sps = 8
SNR_dB = 10
rolloff = 0.35
span = 8

bits = np.random.randint(0, 2, N)

# BPSK:
# 0 -> -1
# 1 -> +1
symbols = 2 * bits - 1


# ============================================================
# 2. ROOT RAISED COSINE FILTER
# ============================================================

def rrc_filter(alpha, span, sps):

    t = np.arange(
        -span/2,
        span/2 + 1/sps,
        1/sps
    )

    h = np.zeros(len(t))

    for i, ti in enumerate(t):

        # t = 0
        if abs(ti) < 1e-10:

            h[i] = (
                1 - alpha
                + 4 * alpha / np.pi
            )

        # t = +/- T/(4*alpha)
        elif (
            alpha > 0
            and abs(abs(ti) - 1/(4*alpha)) < 1e-10
        ):

            h[i] = (
                alpha / np.sqrt(2)
            ) * (
                (1 + 2/np.pi)
                * np.sin(np.pi/(4*alpha))
                +
                (1 - 2/np.pi)
                * np.cos(np.pi/(4*alpha))
            )

        # General case
        else:

            numerator = (
                np.sin(
                    np.pi * ti * (1 - alpha)
                )
                +
                4 * alpha * ti
                * np.cos(
                    np.pi * ti * (1 + alpha)
                )
            )

            denominator = (
                np.pi * ti
                * (1 - (4 * alpha * ti)**2)
            )

            h[i] = numerator / denominator

    # Normalize filter energy
    h = h / np.sqrt(np.sum(h**2))

    return h


# Create RRC filter
h = rrc_filter(
    rolloff,
    span,
    sps
)


# ============================================================
# 3. UPSAMPLING
# ============================================================

upsampled = np.zeros(N * sps)

upsampled[::sps] = symbols


# ============================================================
# 4. TRANSMIT RRC FILTER
# ============================================================

tx_signal = np.convolve(
    upsampled,
    h
)


# ============================================================
# 5. ADD AWGN
# ============================================================

signal_power = np.mean(
    tx_signal**2
)

SNR_linear = 10**(
    SNR_dB / 10
)

noise_power = (
    signal_power / SNR_linear
)

noise = np.sqrt(
    noise_power
) * np.random.randn(
    len(tx_signal)
)

rx_signal = tx_signal + noise


# ============================================================
# 6. MATCHED FILTER
# ============================================================

matched_filter = h[::-1]

mf_output = np.convolve(
    rx_signal,
    matched_filter
)


# ============================================================
# 7. DELAY CALCULATION
# ============================================================

filter_delay = (
    span * sps // 2
)

total_delay = (
    2 * filter_delay
)

print("=" * 60)
print("MANDATORY DELAY VALIDATION")
print("=" * 60)

print(
    f"RRC filter delay       : "
    f"{filter_delay} samples"
)

print(
    f"Matched filter delay   : "
    f"{filter_delay} samples"
)

print(
    f"Total cascade delay   : "
    f"{total_delay} samples"
)

print(
    "\nSymbol detection is performed "
    "only after removing the total delay."
)


# ============================================================
# 8. DELAY COMPENSATION
# ============================================================

sample_indices = (
    total_delay
    + np.arange(N) * sps
)

valid = (
    sample_indices < len(mf_output)
)

sample_indices = sample_indices[
    valid
].astype(int)

detected_samples = mf_output[
    sample_indices
]


# ============================================================
# 9. BPSK DETECTION
# ============================================================

detected_bits = (
    detected_samples >= 0
).astype(int)

BER = np.mean(
    detected_bits
    != bits[:len(detected_bits)]
)


print("\n" + "=" * 60)
print("BPSK DETECTION")
print("=" * 60)

print(
    f"Detected symbols : "
    f"{len(detected_bits)}"
)

print(
    f"BER at {SNR_dB} dB : "
    f"{BER:.6f}"
)


# ============================================================
# 10. EYE DIAGRAM FUNCTION
# ============================================================

def plot_eye(
    signal,
    sps,
    delay,
    num_traces=100,
    title="Eye Diagram"
):

    signal = signal[delay:]

    plt.figure(
        figsize=(9, 5)
    )

    eye_length = 2 * sps

    for i in range(num_traces):

        start = i * sps
        end = start + eye_length

        if end <= len(signal):

            segment = signal[
                start:end
            ]

            time = (
                np.arange(eye_length)
                / sps
            )

            plt.plot(
                time,
                segment,
                linewidth=0.7
            )

    plt.axvline(
        1,
        linestyle="--"
    )

    plt.title(title)
    plt.xlabel("Time (Symbols)")
    plt.ylabel("Amplitude")
    plt.grid(True)

    plt.show()


# ============================================================
# 11. EYE HEIGHT FUNCTION
# ============================================================

def calculate_eye_height(
    signal,
    sps,
    delay
):

    signal = signal[delay:]

    samples = signal[
        sps//2::sps
    ]

    positive = samples[
        samples >= 0
    ]

    negative = samples[
        samples < 0
    ]

    if (
        len(positive) == 0
        or len(negative) == 0
    ):
        return 0

    upper = np.median(
        positive
    )

    lower = np.median(
        negative
    )

    return upper - lower


# ============================================================
# 12. TRANSMITTED SIGNAL
# ============================================================

plt.figure(figsize=(11, 4))

plt.plot(
    tx_signal[:400],
    linewidth=1.2
)

plt.title(
    "Transmitted BPSK Signal after RRC Pulse Shaping"
)

plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()


# ============================================================
# 13. MATCHED FILTER OUTPUT
# ============================================================

plt.figure(figsize=(11, 4))

plt.plot(
    mf_output[:500],
    linewidth=1.2
)

plt.title(
    "Matched Filter Output"
)

plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()


# ============================================================
# 14. DETECTED SYMBOLS
# ============================================================

plt.figure(figsize=(11, 4))

plt.stem(
    detected_samples[:50],
    basefmt=" "
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "Detected BPSK Symbols after Delay Compensation"
)

plt.xlabel("Symbol Number")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()


# ============================================================
# 15. MAIN EYE DIAGRAM
# ============================================================

plot_eye(
    mf_output,
    sps,
    total_delay,
    100,
    "Eye Diagram after Matched Filtering"
)


# ============================================================
# 16. EYE DIAGRAMS FOR DIFFERENT SNR
# ============================================================

SNR_values = [
    0,
    5,
    10,
    15,
    20
]

eye_heights = []
BER_SNR = []

plt.figure(
    figsize=(12, 10)
)

for p, snr in enumerate(
    SNR_values
):

    SNR_linear = (
        10**(snr / 10)
    )

    noise_power = (
        signal_power
        / SNR_linear
    )

    noise = (
        np.sqrt(noise_power)
        * np.random.randn(
            len(tx_signal)
        )
    )

    rx_temp = (
        tx_signal + noise
    )

    mf_temp = np.convolve(
        rx_temp,
        matched_filter
    )

    signal_eye = (
        mf_temp[total_delay:]
    )

    plt.subplot(
        3,
        2,
        p + 1
    )

    for i in range(50):

        start = i * sps
        end = start + 2 * sps

        if end <= len(signal_eye):

            plt.plot(
                np.arange(2*sps) / sps,
                signal_eye[
                    start:end
                ],
                linewidth=0.7
            )

    plt.axvline(
        1,
        linestyle="--"
    )

    plt.title(
        f"SNR = {snr} dB"
    )

    plt.xlabel(
        "Time (Symbols)"
    )

    plt.ylabel(
        "Amplitude"
    )

    plt.grid(True)


    # Eye height
    height = calculate_eye_height(
        mf_temp,
        sps,
        total_delay
    )

    eye_heights.append(
        height
    )


    # BER
    samples = mf_temp[
        sample_indices
    ]

    detected = (
        samples >= 0
    ).astype(int)

    ber = np.mean(
        detected
        != bits[:len(detected)]
    )

    BER_SNR.append(
        ber
    )


plt.suptitle(
    "Eye Diagrams for Different SNR Values",
    fontsize=14
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. EYE HEIGHT VS SNR
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    SNR_values,
    eye_heights,
    "-o",
    linewidth=1.5
)

plt.title(
    "Eye Height versus SNR"
)

plt.xlabel(
    "SNR (dB)"
)

plt.ylabel(
    "Eye Height"
)

plt.grid(True)

plt.show()


# ============================================================
# 18. BER VS SNR
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.semilogy(
    SNR_values,
    BER_SNR,
    "-o",
    linewidth=1.5
)

plt.title(
    "BER versus SNR"
)

plt.xlabel(
    "SNR (dB)"
)

plt.ylabel(
    "BER"
)

plt.grid(True)

plt.show()


# ============================================================
# 19. BER VS TIMING OFFSET
# ============================================================

timing_offsets = np.arange(
    -3,
    4
)

BER_timing = []

for offset in timing_offsets:

    indices = (
        total_delay
        + offset
        + np.arange(N) * sps
    )

    valid = (
        (indices >= 0)
        &
        (indices < len(mf_output))
    )

    indices = indices[
        valid
    ].astype(int)

    samples = mf_output[
        indices
    ]

    detected = (
        samples >= 0
    ).astype(int)

    ber = np.mean(
        detected
        != bits[:len(detected)]
    )

    BER_timing.append(
        ber
    )


plt.figure(
    figsize=(8, 5)
)

plt.semilogy(
    timing_offsets,
    BER_timing,
    "-o",
    linewidth=1.5
)

plt.title(
    "BER versus Sampling Timing Offset"
)

plt.xlabel(
    "Timing Offset (samples)"
)

plt.ylabel(
    "BER"
)

plt.grid(True)

plt.show()


# ============================================================
# 20. ROLL-OFF FACTOR COMPARISON
# ============================================================

rolloff_values = [
    0.1,
    0.25,
    0.5,
    0.75,
    1.0
]

plt.figure(
    figsize=(10, 6)
)

for alpha in rolloff_values:

    h_temp = rrc_filter(
        alpha,
        span,
        sps
    )

    tx_temp = np.convolve(
        upsampled,
        h_temp
    )

    mf_temp = np.convolve(
        tx_temp,
        h_temp[::-1]
    )

    delay_temp = (
        span * sps
    )

    indices = (
        delay_temp
        + np.arange(N) * sps
    )

    valid = (
        indices < len(mf_temp)
    )

    indices = indices[
        valid
    ].astype(int)

    samples = mf_temp[
        indices
    ]

    samples = (
        samples
        / np.max(
            np.abs(samples)
        )
    )

    plt.plot(
        samples[:100],
        label=f"α = {alpha}"
    )


plt.title(
    "Effect of RRC Roll-off Factor"
)

plt.xlabel(
    "Symbol Number"
)

plt.ylabel(
    "Normalized Amplitude"
)

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 21. MULTIPATH ISI CHANNEL
# ============================================================

# Direct path + delayed echo
channel = np.array([
    1.0,
    0,
    0,
    0,
    0.5
])

multipath_signal = np.convolve(
    tx_signal,
    channel
)


# Add AWGN
multipath_power = np.mean(
    multipath_signal**2
)

noise_power = (
    multipath_power
    / 10**(SNR_dB / 10)
)

noise = (
    np.sqrt(noise_power)
    * np.random.randn(
        len(multipath_signal)
    )
)

multipath_rx = (
    multipath_signal
    + noise
)


# ============================================================
# 22. MATCHED FILTER AFTER ISI
# ============================================================

multipath_mf = np.convolve(
    multipath_rx,
    matched_filter
)


# ============================================================
# 23. DETECTION WITH ISI
# ============================================================

indices_isi = (
    total_delay
    + np.arange(N) * sps
)

valid = (
    indices_isi
    < len(multipath_mf)
)

indices_isi = indices_isi[
    valid
].astype(int)

samples_isi = (
    multipath_mf[
        indices_isi
    ]
)

detected_isi = (
    samples_isi >= 0
).astype(int)

BER_ISI = np.mean(
    detected_isi
    != bits[:len(detected_isi)]
)


print("\n" + "=" * 60)
print("MULTIPATH ISI RESULT")
print("=" * 60)

print(
    f"BER without multipath ISI : "
    f"{BER:.6f}"
)

print(
    f"BER with multipath ISI    : "
    f"{BER_ISI:.6f}"
)


# ============================================================
# 24. EYE DIAGRAM WITH MULTIPATH ISI
# ============================================================

plot_eye(
    multipath_mf,
    sps,
    total_delay,
    100,
    "Eye Diagram with Multipath ISI"
)


# ============================================================
# 25. BEFORE / AFTER ISI COMPARISON
# ============================================================

plt.figure(
    figsize=(12, 8)
)


# Without ISI
plt.subplot(
    2,
    1,
    1
)

signal_eye = (
    mf_output[
        total_delay:
    ]
)

for i in range(50):

    start = i * sps
    end = start + 2*sps

    if end <= len(signal_eye):

        plt.plot(
            np.arange(2*sps)/sps,
            signal_eye[
                start:end
            ],
            linewidth=0.7
        )

plt.axvline(
    1,
    linestyle="--"
)

plt.title(
    "Eye Diagram without Multipath ISI"
)

plt.xlabel(
    "Time (Symbols)"
)

plt.ylabel(
    "Amplitude"
)

plt.grid(True)


# With ISI
plt.subplot(
    2,
    1,
    2
)

signal_eye = (
    multipath_mf[
        total_delay:
    ]
)

for i in range(50):

    start = i * sps
    end = start + 2*sps

    if end <= len(signal_eye):

        plt.plot(
            np.arange(2*sps)/sps,
            signal_eye[
                start:end
            ],
            linewidth=0.7
        )

plt.axvline(
    1,
    linestyle="--"
)

plt.title(
    "Eye Diagram with Multipath ISI"
)

plt.xlabel(
    "Time (Symbols)"
)

plt.ylabel(
    "Amplitude"
)

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 26. FINAL RESULTS
# ============================================================

print("\n" + "=" * 60)
print("EXPERIMENT 8 - FINAL RESULTS")
print("=" * 60)

print(
    f"Number of symbols      : {N}"
)

print(
    f"Samples/symbol         : {sps}"
)

print(
    f"RRC roll-off factor    : {rolloff}"
)

print(
    f"Filter span            : {span} symbols"
)

print(
    f"SNR                    : {SNR_dB} dB"
)

print("-" * 60)

print(
    f"Transmit filter delay  : "
    f"{filter_delay} samples"
)

print(
    f"Matched filter delay   : "
    f"{filter_delay} samples"
)

print(
    f"Total cascade delay    : "
    f"{total_delay} samples"
)

print("-" * 60)

print(
    f"BER without ISI        : "
    f"{BER:.6f}"
)

print(
    f"BER with multipath ISI : "
    f"{BER_ISI:.6f}"
)

print("=" * 60)
print("EXPERIMENT 8 COMPLETED")
print("=" * 60)
```
